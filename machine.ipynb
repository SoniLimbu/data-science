        error caused by over simple assumptions in a model

Stadard scaler -one of the techniques used in preprocessing

```python
from sklearn.preprocessing import StandardScaler
X = [
    [20,30000],
    [25,5000],
    [30,40000], 
     ]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
from sklearn.preprocessing import MinMaxScaler
X = [
    [20,30000],
    [25,5000],
    [30,40000], 
     ]

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

Encoding - numerical format 
1.label encoding -male =0 ,female -1 -
2.ordinal encoding : levels of data low medium high 
3.one hot encoder -matrix format convert data 

```python
from sklearn.preprocessing import LabelEncoder

gender = ['Male', 'Female', 'Female', 'Male', 'Female']
encoder = LabelEncoder()
gender_encoded = encoder.fit_transform(gender)
print(gender_encoded)
```

```python
print(encoder.classes_)
```

odinal encoding is a technique used to convert categorical variables into numerical values. In this case, the `LabelEncoder` from the `sklearn.preprocessing` module is used to encode the gender variable. The unique classes (Male and Female) are assigned integer values (0 and 1) respectively. The output of `encoder.classes_` shows the original classes in the order they were encoded.

```python
from sklearn.preprocessing import OrdinalEncoder

size = [['Small'], ['Medium'], ['Large'], ['Medium'], ['Small']]
encoder = OrdinalEncoder()
encoded_size = encoder.fit_transform(size)
encoder.fit_transform(size)
print(encoded_size)
```
