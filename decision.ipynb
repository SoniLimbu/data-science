## Decision Tree Algorithm:
1.root node :
        first /top node of tree

2.Internal Node : a node where the model ask another question ans split thedata .

Decision tree :
                -ID3 -iterative dichotomiser 3 
                -which feature is the best one to split the data?
                -it uses entropy and Information gain


        steps:
        1.calcualte Entropy of entire dataset 
        2.calculate information gain for each dataset 
        3.chose feature with higher information gain
        4.make it root node 
        5.split dataset
        6.Repeat the process for each branch
        7.stop when appropriate 


entropy:measure how mixed /impure the classes are:

        -if all classes are same then entropy is 0
        -if classes are equally mixed then entropy is 1


information Gain:
-how much did the uncertaininty decrease after making a split 

IG  = Entropy(parent) - weighted entropy 


eg;

age = 0.10
