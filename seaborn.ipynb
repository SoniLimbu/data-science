```python
import seaborn as sns
dataset_names = sns.get_dataset_names()
print(dataset_names)
```

```python
import matplotlib.pyplot as plt

sns.scatterplot(data=df,x="total_bill",y="tip")
plt.title("total bill vs Tip")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)
plt.title("total bill vs Tip by sex")
plt.show()
```

```python
#box plot 

sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)
plt.title("total bill by day")
plt.show()
```

Countplot:count how many times each category apperars in the datasets.

```python
df['day'].unique()
```

```python
sns.countplot(
    data=df,
    x='day'
)
plt.title("Number of customers by day")
plt.show()
```

```python
sns.histplot(data=df,x="total_bill")

plt.title("distribution of total bill")
plt.xlabel("Total bill")
plt.ylabel("count")

plt.show()
```

```python
sns.pairplot(df)
plt.show()
```

```python
corr =df.corr(numeric_only=True)
sns.heatmap(
    corr,
    annot=True,
    vmin=-1,
    vmax=1
)
plt.title("corealation matrix")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
corr= corr.where(~corr.eq(1))

print(corr.unstack().sort_values(ascending=False).head())
```

```python
sns.boxplot(data=df,x="total_bill")
```

```python
sns.boxplot(data=df)
plt.title("outlier of numerical columns")
plt.show()
```

```python
categorical_df = df.select_dtypes(exclude="number")

for i in categorical_df.columns:
    sns.countplot(data=df,x=i)
    plt.title(f"count of {i}")
    plt.show()
    
```
