Matplotlib is a low level graph plotting library in python that serves as a visualization utility.

Matplotlib was created by John D. Hunter.

Matplotlib is mostly written in python, a few segments are written in C, Objective-C and Javascript for Platform compatibility.



Matplotlib is a popular Python data visualization library used to create static, animated, and interactive plots. It is especially powerful for 2D plotting, similar to what MATLAB offers, and is widely used in data science, machine learning, and scientific computing.

Matplotlib mainly revolves around two modules:

1.pyplot: A collection of functions that make Matplotlib work like MATLAB.

2.Figure and Axes objects: For fine-grained control over the layout.




1.Lineplot :
.plot(X_data,y_data)

```python
import matplotlib.pyplot as plt 
X = [0,2,5,9,6]
y = [2,8,9,6,3]
plt.plot(X,y,marker="o",color="GREEN")
plt.xlabel("X-axis label")
plt.ylabel("Y-axis label")
plt.title("Plot Title")
plt.grid(True,linestyle='--',alpha=1)
plt.show()

```

## Bar chart 
-display categorical data using rectangular bars whose length are proportional to the values they represent
plt.bar()

```python
students = ["Ram","balen","rabi","hari"]

marks = [70,85,96,90]
plt.bar(
    students,
    marks,
    width=0.6,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8
)
plt.xlabel("students")
plt.ylabel("marks")
plt.title("students Marks")
plt.grid(axis="y",linestyle="--",alpha=0.5)
plt.show()
```

```python
subjects = ["Python","Java","R"]
marks = [78,85,96]
plt.bar(
    subjects,
    marks,
    width=0.6,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8,
    label="Marks"
)
plt.xlabel("students")
plt.ylabel("marks")
plt.title("students Marks")
plt.grid(axis="y",linestyle="--",alpha=0.5)

#set y-axis range 
plt.ylim(0,200)

for i , value in enumerate(marks):
    plt.text(i,value + 1,str(value),ha="center")
    
plt.legend()
plt.show()
```

Histogram:
statistical diagram -shows the distribution of numerical data .
-bins
it tell us how many data values fall inside different numerical ranges(interval/bins).
histogram = numerical data + bins + frequency 

```python
marks = [70,80,89,74,75,55,66,55,89,57,65,96,89,68,96,48,42,43,44,54,65,68,69]
plt.hist(
    marks ,
    bins=10, #important parameter is bins , how much data range into 5 intervals 
    color="skyblue",
    edgecolor="black",
    alpha=0.8,
    
)
plt.xlabel("marks")
plt.ylabel("Frequency")
plt.title("distribution of student")
plt.show()
```

4.Scatter plot :
it is used to see the relationship between two numerical variables 
upward trend - positive relationship
downward trend - negative relationship
randomly scaterred - weak / no relationship

```python
hours = [1,2,3,4,5,6,7]
marks = [45,50,55,62,68,75,82]
plt.scatter (
    hours,
    marks,
    s=100,
    color="skyblue",
    edgecolors="black",
    alpha=0.8,
    marker="o"
)
plt.xlabel("hours studied")
plt.ylabel("marks")
plt.title("study hours vs  Exam marks")

plt.show()
```

## Pie-chart:
represents how total is divided into different categories or proportions.


```python
categories = ["food","Travel","study","Entertainmet"]
expenses = [ 4000,2000,3000,1000]
plt.pie(
    expenses,
    labels = categories,
    autopct="%1.01f%%",
    startangle=90,
    shadow=False
)
plt.title("montly expenses")
plt.show()
```
