```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
        self.total = 0
        self.average = 0
        self.grade =" "
    
    def calculate_total_average(self):
        self.total = sum(self.marks)
        self.average = self.total / len(self.marks)
        
    def assign_grade(self):
        if self.average >=90:
            self.grade = "A"
        elif self.average >=80:
            self.grade ="B"
        elif self.average >=70:
            self.grade ="C"
        elif self.average >= 60:
            self.grade = "D"
        else:
            self.grade ="F"
            
            
    def show_result(self):
        print(f"Student name is {self.name}")
        print(f"Student marks is {self.marks}")
        print(f"Student total marks is {self.total}")
        print(f"Student Average marks is {self.average:.2f}")
        print(f"Grade is :{self.grade}")
        
        
name = input("enter the student name")
marks=[]
for i in range(1,4):
    mark=float(input("enter marks of subject"))
    marks.append(mark)
    
student = Student(name,marks)
student.calculate_total_average()
student.assign_grade()
student.show_result()
```

Create a parent class MLmodel with model name and accuracy and a method checkperformace that prints whether the model has a good performance.
Then again create a child class knn with additional attribute k
create a method show_details in knn that display the model name accuracy and value of k

```python
class MLmodel:
    def __init__(self, model_name, accuracy):
        self.model_name = model_name
        self.accuracy = accuracy

    def checkperformace(self):
        if self.accuracy >= 0.80:
            print("The model has good performance.")
        else:
            print("The model performance needs improvement.")


class knn(MLmodel):
    def __init__(self, model_name, accuracy, k):
        super().__init__(model_name, accuracy)
        self.k = k

    def show_details(self):
        print(f"Model name: {self.model_name}")
        print(f"Accuracy: {self.accuracy:.2%}")
        print(f"Value of k: {self.k}")


knn_model = knn("K-Nearest Neighbors", 0.90, 5)
knn_model.show_details()
knn_model.checkperformace()
```
