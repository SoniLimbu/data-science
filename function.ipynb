check whether n is divisible by i 


```python
def is_prime(n):
    if n  <=1:
        return False
    
    for i in range(2,n): #(2,7)
        if n% i ==0:
            return False
    return True 

num = int(input("enter a number"))
if is_prime(num):
    print(num, "is prime")
else:
    print(num,"is not prime")
```

Write a function reverse_string that takes a string as input and return the string in reverse order

```python
def reverse_string(text):
    return text[::-1]

print(reverse_string("Thankyou"))
```

```python
def calculate_marks(numbers):
    total = sum(numbers)
    mean = total / len(numbers)
    return mean

numbers = [10,20,30,80,10]
result= calculate_marks(numbers)
print(f"Mean is {result}")
```

lambda functions:
lambda fn are anonymous functions defined using the lambda keyword.
syntax:
lambda argument : expression
add of two 

lambda function:


```python
def double(n):
    return n*2

print(double(10))
```

```python
double = lambda n:n*2   #n means arguments or syntax 
print(double(10))
```

```python
square = lambda n:n*n
number = float(input())
result = square(number)
print(result)
```

#normalizing a single feature value 


```python
age = 40 
max_age =100
normalized_age = lambda x , maximum :x /maximum
normalized_age = normalized_age(age,max_age)
print("Normalized age", normalized_age)
```
