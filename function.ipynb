```python
def is_prime(n):
    if n  <=1:
        return False
    
    for i in range(2,n): #(2,7)
        if n% i ==0:
            return False
    return True 

num = int(input("enter a number"))
if is_prime(num):
    print(num, "is prime")
else:
    print(num,"is not prime")
```

Write a function reverse_string that takes a string as input and return the string in reverse order

```python
def reverse_string(text):
    return text[::-1]

print(reverse_string("Thankyou"))
```

```python
def calculate_marks(numbers):
    total = sum(numbers)
    mean = total / len(numbers)
    return mean

numbers = [10,20,30,80,10]
result= calculate_marks(numbers)
print(f"Mean is {result}")
```

```python
operations = {
    'add': lambda x, y: x + y,
    'subtract': lambda x, y: x - y,
    'multiply': lambda x, y: x * y
}

result_add = operations['add'](5, 3)
print(f"Addition: {result_add}")

result_subtract = operations['subtract'](10, 4)
print(f"Subtraction: {result_subtract}")

result_multiply = operations['multiply'](10, 4)
print(f"multiplication: {result_multiply}")
```

#list comprehension:
it is short and easy way to create a new list from an existing sequence such as list , range or string 
syntax:
newlist = [expression for item in iterable]

```python
numbers = [1,2,3,4,5]
square = []
for n in numbers:
    square.append(n*n)

print(square)
```

```python
numbers = [1,2,3,4,5]
square= [n*n for i in numbers]
print(square)
```

```python
names = ["soni","limbu"]
upper= [name.upper() for name in names]
print(upper)
```

```python
marks =[]
for i in range(5):
    mark = int(input("enter the marks :"))
    marks.append(mark)

passed = [mark for mark in marks if mark>=40] ## list comprehension
print("all marks",marks)
print("passing marks",passed)
```

Ask the user to enter the marks of  5 student and assign grades:
80 or above -A
60-79-B
40-59-B-
below 40 -Fail

```python
marks = []

for i in range(5):
    mark = int(input("Enter the marks: "))
    marks.append(mark)

grades = ["A" if mark >= 80 else 
          "B" if mark >= 60 else 
          "B-" if mark >= 40 else
          "Fail" for mark in marks]

print("All marks:", marks)
print("Grades:", grades)
```
