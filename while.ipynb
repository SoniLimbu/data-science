Write a python that continuously takes numbers from the user and add them together .the program must halt when the user enters 0 and print its sum

```python
total = 0
num = int(input("enter the number "))

while num!=0:
    total = total + num
    num = int(input("enter the numbers"))

print("total of the number is ",total)
```
