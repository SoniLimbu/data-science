Exception Handling:
- it is the process of responding to exceptions in a custom way during the exectuion of a program.
syntax:
try:
#code that may cause an exception 

except:
#code to run when exception occurs

```python
try:
   numerator = int (input("enter the numerator:"))
   denominator = int(input("enter the denominator:"))
   result = numerator/denominator
   print(result)

except:
    print("denominator cannot be 0 try again")
```

```python
try:
    numbers = [10,20,30]
    index = int(input())
    print(numbers[index])
    
except:
    print("index is wrong")
    
finally:
    print("always printed")
```

```python
students = {"soni","arati","smita"}
try:
    name= input("enter the students name to remove")
    students.remove(name)
    print("updated set", students)
except:
    print("student not found in the set")
finally:
    print("operation completed")
    
```

```python
try:
    age = int(input("enter your age"))
    
    if age <0:
        raise ValueError("age cannot be 0 or negative")
    
    if age >18:
        predicition = "adult"
        
    else:
        predicition ="minor"
    print("model prediction", predicition)
    
except ValueError:
    print("invalid age has been entered")

finally:
    print ("prediction process completed")
    
    

    
        
```
