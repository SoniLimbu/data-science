1. E-commerce cart validation A shopping cart is stored as a list of product IDs, and a second list contains product IDs that are out of stock. Using only Python list/tuple operations, create the list of cart items that can actually be purchased while preserving their original order. Do not use loops or set operations.

```python
cart = ['p101','p102','p103','p104']
out_of_stock = ['102','104']
purchased = cart.copy()
 
if out_of_stock[0] in purchased:
    purchased.remove(out_of_stock[0])
if out_of_stock[1] in purchased:
    purchased.remove(out_of_stock[1])

print("purchased items:", purchased)
    
```

2. Student score analysis A school stores exam scores in a list such as [78, 91, 65, 91, 84, 73, 65, 88]. Using list/tuple operations only, determine the highest score, lowest score, total score, average score, and the number of times the highest score occurs. Avoid loops.

```python
school=[78, 91, 65, 91, 84, 73, 65, 88]
highest_score = max(school)
lowest_score = min(school)
total_score = sum(school)
average_score = total_score / len(school)
highest_count = school.count(highest_score)

print("Highest score:", highest_score)
print("Lowest score:", lowest_score)
print("Total score:", total_score)
print("Average score:", average_score)
print("Number of students with the highest score:", highest_count,"times")
```

3. A bank transaction record is stored as a list of signed amounts, for example [1200, -250, -100, 500, -75, 300]. Using list/tuple operations only, separate the deposits and withdrawals into two lists, then calculate the total deposited and total withdrawn. Do not use loops.

```python
transactions = [1200, -250, -100, 500, -75, 300]

deposits = [transactions[0], transactions[3], transactions[5]]
withdrawals = [transactions[1], transactions[2], transactions[4]]

total_deposited = sum(deposits)
total_withdrawn = sum(withdrawals)

print("Deposits:", deposits)
print("Withdrawals:", withdrawals)
print("Total deposited:", total_deposited)
print("Total withdrawn:", abs(total_withdrawn))
```

4.A restaurant receives an order as a list of item tuples: [('Burger', 2), ('Fries', 1), ('Burger', 1), ('Juice', 2)]. 5.Available airline seats are represented by a list of seat labels such as ['12A', '12B', '12C', '14A', '14B', '15A']. A passenger requests a tuple of preferred seats, e.g. ('14B', '13A', '12C'). 18.A stock's recent closing prices are stored as a list: [184.2, 186.5, 183.9, 190.1, 188.7, 192.4]. Using list/tuple operations only, calculate the highest close, lowest close, price range, number of prices, and create a tuple containing the first, middle, and last prices. Do not use loops.

```python
prices = [184.2, 186.5, 183.9, 190.1, 188.7, 192.4]

highest = max(prices)
lowest = min(prices)
price_range = highest - lowest
number_of_prices = len(prices)

first_middle_last = (
    prices[0],
    prices[len(prices) // 2],
    prices[-1]
)

print("Highest:", highest)
print("Lowest:", lowest)
print("Price range:", price_range)
print("Number of prices:", number_of_prices)
print("First, middle, last:", first_middle_last)
```

20.A logistics system stores shipments as nested tuples: [('NP01', ('Kathmandu', 'Pokhara'), 120), ('NP02', ('Biratnagar', 'Dharan'), 80), ('NP03', ('Lalitpur', 'Butwal'), 150)]. Using only list/tuple indexing, slicing, count/index, and built-in functions, extract the route of NP02, find the largest shipment weight, identify the shipment record containing that weight, and create a tuple containing the first and last shipment IDs. Do not use loops, sets, dictionaries, or comprehensions.

```python
shipments = [
    ('NP01', ('Kathmandu', 'Pokhara'), 120),
    ('NP02', ('Biratnagar', 'Dharan'), 80),
    ('NP03', ('Lalitpur', 'Butwal'), 150)
]

# Route of NP02
np02_route = shipments[1][1]

# Largest shipment weight
largest_weight = max(
    shipments[0][2],
    shipments[1][2],
    shipments[2][2]
)

# Shipment record containing largest weight
if shipments[0][2] == largest_weight:
    largest_record = shipments[0]
elif shipments[1][2] == largest_weight:
    largest_record = shipments[1]
else:
    largest_record = shipments[2]

# First and last shipment IDs
shipment_ids = (shipments[0][0], shipments[-1][0])

print("NP02 route:", np02_route)
print("Largest weight:", largest_weight)
print("Largest shipment:", largest_record)
print("First and last IDs:", shipment_ids)
```
