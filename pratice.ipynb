```python
#6
def min_max(numbers):
    return (min(numbers), max(numbers))


numbers = [10, 5, 20, 3, 15]

minimum, maximum = min_max(numbers)

print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
#12
products = {
    "Fruits": [100, 120, 80],
    "Vegetables": [50, 60, 70],
    "Drinks": [100, 150, 120]
}

for category, prices in products.items():
    average = sum(prices) / len(prices)
    print(category, "average:", average)
```

```python
#13
january = {
    "Laptop": 5,
    "Mouse": 10,
    "Keyboard": 7
}

february = {
    "Laptop": 3,
    "Mouse": 8,
    "Monitor": 4
}

total_sales = january.copy()

for product, units in february.items():
    if product in total_sales:
        total_sales[product] += units
    else:
        total_sales[product] = units

print(total_sales)
```

```python
#14
employees = {
    101: "Soni",
    102: "Ram",
    103: "Anita"
}

inverted = {}

for employee_id, name in employees.items():
    inverted[name] = employee_id

print(inverted)
```

```python
#15
math_students = ["Soni", "Ram", "Anita", "Hari"]
science_students = ["Anita", "Hari", "Rita", "John"]

math_set = set(math_students)
science_set = set(science_students)

both = math_set & science_set

print(both)
```

```python
#16
math_students = ["Soni", "Ram", "Anita", "Hari"]
science_students = ["Anita", "Hari", "Rita", "John"]

math_set = set(math_students)
science_set = set(science_students)

all_students = math_set | science_set

print(all_students)
```
