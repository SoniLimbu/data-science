#7. Write a program that stores 10 in x and 20 in y, then swaps their values WITHOUT using a third (temporary) variable. Print the values before and after the swap.

```python
x = 10
y = 20

print("Before swap:")
print("x =", x)
print("y =", y)

x, y = y, x

print("After swap:")
print("x =", x)
print("y =", y)
```

#11. Write a program to calculate and print the area and perimeter of a rectangle whose length is 12 and width is 5.
